# 1. Umieszczam dane w Digital Ocean Spaces

W tym notebooku wysyłam oba pliki z wynikami półmaratonu wrocławskiego (2023 i 2024) do mojego bucketa w Spaces.
Robię to tak jak w lekcji `10__cloud_storage.ipynb` (biblioteka `boto3`).

````python
import os
import boto3
from dotenv import load_dotenv

load_dotenv()
````

````python
# Klient tworzę jak w lekcji 10 - klucze i endpoint boto3 bierze sam ze zmiennych środowiskowych z .env
s3 = boto3.client("s3")
````

````python
BUCKET_NAME = "jerry-77"
# Dane trzymam pod osobnym prefixem, żeby nie mieszały się z plikami z lekcji
DATA_PREFIX = "polmaraton/data/"

FILES = [
    "halfmarathon_wroclaw_2023__final.csv",
    "halfmarathon_wroclaw_2024__final.csv",
]
````

````python
# Wysyłam oba pliki do Spaces (upload_file z lekcji 10)
for file in FILES:
    s3_path = f"{DATA_PREFIX}{file}"
    print(f"Wysyłam plik {file} do {s3_path}")
    s3.upload_file(
        Filename=file,
        Bucket=BUCKET_NAME,
        Key=s3_path,
    )
````

````python
# Sprawdzam, czy pliki są w buckecie - listowanie po prefixie z lekcji 10
response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix=DATA_PREFIX)
for obj in response["Contents"]:
    print(obj["Key"], obj["Size"])
````
